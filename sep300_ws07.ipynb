```python
%%writefile task1.py

# Task 1

from data import items
import sqlite3  # SQLite databases

# open connection -- store.db
conn = sqlite3.connect("store.db")

# cursor -- run SQL command
cursor = conn.cursor()

cursor.execute("DROP TABLE IF EXISTS products")

cursor.execute("""

    CREATE TABLE products (
    
        product_name        CHAR(20) NOT NULL,
        price               NUMERIC(10,2) NOT NULL,
        quantity_in_stock    INTEGER NOT NULL,
        expiry_date         DATE
    
    )

""")


for item in items:
    # only date has none
    if item["expiry_date"] is None:
        expiry = "NULL"

    else:
        expiry = f"'{item['expiry_date']}'"

    cursor.execute(
        f"INSERT INTO products (product_name, price, quantity_in_stock, expiry_date) VALUES ('{item['product_name']}', {float(item['price'])}, {int(item['quantity_in_stock'])}, {expiry})"
    )

conn.commit()
conn.close()
```

```python
# task 3 -- ORM obj instead of SQL
from datetime import date
from sqlalchemy import create_engine
from sqlalchemy.orm import DeclarativeBase, Mapped, mapped_column, Session


class Base(DeclarativeBase):
    pass


class Product(Base):
    __tablename__ = "products"

    # Mapped[type] -- this attr --> a col of that type
    # ORM needs PK --> name

    product_name: Mapped[str] = mapped_column(primary_key=True)
    price: Mapped[float]
    quantity_in_stock: Mapped[int]

    expiry_date: Mapped[date | None]  # date | None -- can be date or emupty


engine = create_engine("sqlite:///store.db")

# use `with` to close section automaticlly at  the end
with Session(engine) as session:
    print("Items in stock less than 10")
    products = (
        session.query(Product)  # 1. which table
        .filter(Product.quantity_in_stock < 10)  # 2. which rows
        .all()
    )  # 3. run it, get results

    for p in products:
        print(p.product_name, p.price, p.quantity_in_stock, p.expiry_date)
```
